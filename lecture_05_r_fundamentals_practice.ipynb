## 6. Comparison & Logical Operators
* Comparison: `==`, `!=`, `>`, `<`, `>=`, `<=`
* Logical: `&` (AND), `|` (OR), `!` (NOT)

````r
reads <- 1250
mito_pct <- 8.5

# Comparison tests
cat("Reads >= 1000:", reads >= 1000, "\n")

# Compound logical conditions
pass_qc <- (reads >= 500) & (mito_pct < 15.0)
cat("Passes QC (AND):", pass_qc, "\n")

flag_spot <- (reads < 200) | (mito_pct > 20.0)
cat("Flagged spot (OR):", flag_spot, "\n")

cat("Invert flag (NOT):", !flag_spot, "\n")
````

## 7. Control Flow (`if/else`, loops)

````r
# 1. Branching
expr <- 450
if (expr > 500) {
  cat("Tier: High\n")
} else if (expr >= 100) {
  cat("Tier: Medium\n")
} else {
  cat("Tier: Low\n")
}

# 2. for loop with next & break
for (cnt in c(1200, 45, 890, -99, 1400)) {
  if (cnt < 0) {
    cat("Corrupted value (", cnt, ")! Aborting loop.\n")
    break
  }
  if (cnt < 100) {
    next  # skip low count spot
  }
  cat("Processing valid count:", cnt, "\n")
}

# 3. while loop
resolution <- 0.2
while (resolution <= 0.6) {
  cat("Clustering resolution:", resolution, "\n")
  resolution <- resolution + 0.2
}
````

### 📌 Problem 3: Quality Control Audit (Guard Clauses & Tiering)
A spatial sequencing run produces a vector of mitochondrial contamination readings (in percent) across 6 tissue spots:
`mito_readings <- c(4.8, 12.3, 85.0, -1.0, 18.5, 9.2)`

Write a `for` loop that iterates over `mito_readings`:
1. **Sensor Glitch Check**: If reading is negative (`< 0`), print `"Sensor Error ([reading]%)! Aborting audit."` and exit the loop immediately with `break`.
2. **Dead Spot Filter**: If reading is strictly greater than `20.0%` (lysed/dead cell), skip it using `next`.
3. **Tier Passing Spots**:
   * If reading `< 10.0%`, print `"Spot [reading]% mito: Pristine"`
   * If reading `>= 10.0%`, print `"Spot [reading]% mito: Acceptable"`

````r
mito_readings <- c(4.8, 12.3, 85.0, -1.0, 18.5, 9.2)

# --- Problem 3: Your Code Here ---

for (mito in mito_readings) {
  # 1. Break if sensor error (< 0)
  if (mito < 0) {
    cat("Sensor Error (", mito,"%)! Aborting audit.\n", sep = "")
    break
  }
   # 2. Skip if dead spot (> 20.0)
   if (mito > 20.0) {
    next
  }
  # 3. Classify Pristine vs Acceptable
  if (mito < 10) {
    cat("Spot ", mito, "% mito: Pristine\n")
  }
  if (mito >= 10) {
    cat("Spot ", mito, "% mito: Acceptable\n")
  }

}

mito_readings <- c(4.8, 12.3, 85.0, -1.0, 18.5, 9.2)

````
